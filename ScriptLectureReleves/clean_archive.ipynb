```python
import paramiko
from scp import SCPClient
import shutil


def createSSHClient(server, port, user, password):
    client = paramiko.SSHClient()
    client.load_system_host_keys()
    client.set_missing_host_key_policy(paramiko.AutoAddPolicy())
    client.connect(server, port, user, password)
    return client


ssh = createSSHClient('192.168.10.77', 22, 'softia', 'Softia321+')
scp = SCPClient(ssh.get_transport())

scp.put(r'C:\Users\Nabil\Desktop\Softia_Work_Repos\SHUTRefonte\shut-refonte\ScriptLectureReleves\PD3',
        '/home/softia/shut/data', recursive=True)

shutil.rmtree(r'C:\Users\Nabil\Desktop\Softia_Work_Repos\SHUTRefonte\shut-refonte\ScriptLectureReleves\PD3')
```

```python
import os
path = r'C:\Users\Nabil\Desktop\Softia_Work_Repos\SHUTRefonte\DataSHUT'
directories = os.listdir(path)
try:
    for directory in directories:
        items = os.listdir(path + '/' + directory)
        for item in items:
            if item.endswith('.sql'):
                f = open(path + '/' + directory + '/' + item, mode="r")
                print(f.name)
                lines = f.readlines()
                for index, line in enumerate(lines):
                    if line.rfind('SELECT') != -1:
                        lines[index] = ' '
                lines = [x for x in lines if x != ' ']
                f.close()
                f = open(path + '/' + directory + '/' + item, mode="w")
                f.writelines(lines)
                f.close()
except Exception as e:
    print(e)
```
